## What is Git for?

**Git is version control. It keeps your code's history.**

In Lab 03, we saved the seed and inputs. We need the code that used them too.

- **Commit:** a snapshot you choose to record, with a message, author, and links to earlier commits.
- **Compare:** see what changed and where a problem started.
- **Recover:** get back to work you've committed.
- **Branch and merge:** try changes separately, then combine them.
- **Work offline:** a normal clone includes the files and history. Git is *distributed*.

Keep one project with a history, instead of `simulation_final.py`,
`simulation_final2.py`, and `simulation_really_final.py`.

- For a result, record **commit + seed + inputs + environment**.
- **Save, commit, and push are separate steps.**
- Still check the physics. Git doesn't tell us whether the calculation is right.

[More: What is Git?](https://git-scm.com/book/en/v2/Getting-Started-What-is-Git%3F)


### What does GitHub add?

**GitHub hosts Git repositories online.** Git works without it.

| Git | GitHub |
| --- | --- |
| Keeps commits and branches | Hosts the code and history |
| Compares and combines changes | Adds sharing, permissions, and review |
| Works locally, including offline | Lets others access the work online |

- **Pull requests:** propose and review changes.
- **Issues:** track problems and work to do.
- **Forks:** make your own repository from someone else's; contribute back if you want.
- **Actions:** run automated checks.

Here, **local** means the machine running Git: DataHub or your laptop.
DataHub runs on a server; GitHub holds another copy.

You committed offline. Can your partner see it on GitHub yet?

[Browser walkthrough: GitHub Hello World](https://docs.github.com/en/get-started/using-github/hello-world)

2. Save your files. Check for a **clean working tree**; commit work you want to keep before switching or merging.
3. Update from your fork, then check the course:

```bash
git status
git switch main
git pull --ff-only origin main
git fetch upstream
git log --oneline main..upstream/main
```

| Command | What it does |
| --- | --- |
| `pull --ff-only origin main` | Brings in changes from **your GitHub fork** |
| `fetch upstream` | Downloads **course history**; leaves working files alone |
| `log main..upstream/main` | Lists course commits your `main` doesn't have |

`upstream/main` is our local record after the fetch. An empty log means no new course commits.

4. Merge the course history into `main`, then push to your fork:

```bash
git merge --no-edit upstream/main
git log --oneline --graph --all --decorate -10
git push origin main
git status
```

- **Already up to date?** Nothing new to merge. Repeat when new material is posted.
- **Both sides have new commits?** Git may make a merge commit keeping both histories.
- No `--ff-only` on this merge: your homework can make the histories diverge.
- **`--no-edit`** accepts Git's generated merge message.
- **Conflict?** Look at the files with me; the optional exercise below covers this. Don't force-push or discard work.

**`origin` is your fork.** Pulling from it alone won't fetch course updates.
After fetching `upstream`, you still need to **merge**, then **push**.

[More: syncing a fork](https://docs.github.com/en/pull-requests/how-tos/work-with-forks/syncing-a-fork)

### A small command reference

| Question or action | Command |
| --- | --- |
| Where am I? | `pwd` |
| What changed, and what is staged? | `git status` |
| What changed in tracked files since staging? | `git diff` |
| What will the next commit contain? | `git diff --staged` |
| Stage a particular file | `git add filename` |
| Record the staged changes | `git commit -m "Explain the change"` |
| Read the history and branch labels | `git log --oneline --graph --all --decorate` |
| Create and switch to a branch | `git switch -c branch-name` |
| Switch to an existing branch | `git switch main` |
| Merge into the current branch | `git merge branch-name` |
| Check remote URLs | `git remote -v` |
| Download history from my fork | `git fetch origin` |
| Download new course history | `git fetch upstream` |
| Merge course updates into my current branch | `git merge --no-edit upstream/main` |
| Update from my own fork when `main` can fast-forward | `git pull --ff-only origin main` while on `main` |
| Publish `main` | `git push origin main` |
| Publish a new feature branch and track it | `git push -u origin branch-name` |

1. Run `git status` before and after a change.
2. Read the diff. Stage the files you want, then inspect the staged diff.
3. Commit. Push when you're ready to share the commits.
4. If Git stops with an error, read it before trying another command.


### When something stops you

| What you see | What to check |
| --- | --- |
| `not a git repository` | Run `pwd`; enter the clone, not its parent directory. |
| `Author identity unknown` | Set `user.name` and `user.email` inside the repository. |
| `nothing to commit` | Save your file, then check `git status`. Did you already commit the change? |
| `Permission denied (publickey)` when cloning, fetching, or pushing your fork | Check your fork's SSH URL, `ssh-add -l`, and that the public key is on your GitHub account. Repeat the connection test above. |
| No connection to the authentication agent | Rerun the `ssh-agent` and `ssh-add` block in this terminal. |
| Push rejected as non-fast-forward | The remote has history your branch doesn't contain. Fetch, look at the graph, and show me before combining the changes. |
| Switching or merging would overwrite local changes | Look at the changes and commit the work you want to keep before switching. Don't discard it just to get past the error. |
| Conflict markers in a file | Resolve the content, remove markers, stage, inspect, and complete the merge. |

- Save notebooks before staging. Outputs and metadata can make diffs large; clear outputs you don't need.
- Agree on who edits each notebook. Move frequently shared code into small text modules to make changes easier to compare and merge.
- GitHub's notebook preview displays the file; it doesn't run the Python code.
