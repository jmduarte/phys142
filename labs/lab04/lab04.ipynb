## What is Git for?

- **Keep the code, and keep its history.**
- **See what changed between two versions.**

In Lab 03, we kept the seed and inputs so we could repeat a simulation. We also
need the code that used them. If you change the code tomorrow, how do you get
back to the version that produced today's result?

**Git is a version-control system.** It keeps a history of your project as
**commits**: snapshots you choose to record, with a message explaining the change.
Each commit also records an author and links to the commits it came from. We can
compare versions, find where a problem started, and recover work we've committed.
We can also try changes on separate **branches**, then combine them when we're ready.

This lets us keep one project instead of making `simulation_final.py`,
`simulation_final2.py`, and `simulation_really_final.py`. For a simulation, record
the commit along with the seed, parameters, and environment.

Git is **distributed**: a normal clone gives you a local copy of the repository
and its history. You can make commits without being online. Saving a file doesn't
commit it, though, and committing it doesn't upload it. We'll do those steps separately.

Keeping the history doesn't tell us whether the calculation is right. We still
have to check the physics and the results.

For more on how Git stores the history, see [What is Git?](https://git-scm.com/book/en/v2/Getting-Started-What-is-Git%3F).


### What does GitHub add?

- **Put the repository somewhere we can share it.**
- **Review a change before adding it.**

**GitHub hosts Git repositories online.** It gives us a place to share the code
and history, decide who can change them, and discuss contributions. **Pull
requests** let us propose and review changes. **Issues** let us keep track of
problems or work to do. **Forks** let us make our own repository from someone
else's and contribute back. GitHub can also run automated checks through Actions.

You don't need GitHub to use Git. Here we'll use Git on DataHub and keep a copy
on GitHub. DataHub itself runs on a server, but we'll call its files **local**:
they're on the machine where we're running Git.

| Git | GitHub |
| --- | --- |
| Keeps commits and branches | Hosts copies of Git repositories |
| Compares and combines changes | Adds pull requests, review, and discussion |
| Works locally, including offline | Lets us share the work online |

If you make a commit while you're offline, can your partner see it on GitHub?
What still needs to happen?

GitHub's [introductory tutorial](https://docs.github.com/en/get-started/using-github/hello-world)
walks through the browser steps we'll use next.


### Where is the change right now?

- **Edit. Stage. Commit. Push.**

The **working tree** is the set of files you're editing. The **staging area**
holds the content you've selected for the next commit. The **local repository**
keeps the committed history in the hidden `.git` directory. A **remote** is
another repository, usually reached through a URL we've saved.

## Forks: your copy of the course repository

We'll make a **fork** of [the course repository](https://github.com/jmduarte/phys142):
a repository under your own GitHub account that starts from the course history.

Your commits don't change the course repository. Course updates don't

`origin/main` is our **local record** of GitHub's `main` after the fetch. It only
knows what we've fetched; it doesn't watch GitHub for changes. The log and diff
show the new commit, but `cat homework/README.md` still shows the old local file.

Check that the working tree is clean. Now bring that commit into our `main`:

```bash
git merge --ff-only origin/main
cat homework/README.md
git status
```

Next time, `git pull --ff-only origin main` can do the fetch and fast-forward
together in this same situation. We don't need to commit the browser edit again:
we've just brought in the commit that already exists.

`--ff-only` stops if the local and remote histories have diverged. If that happens,

look at the two histories before combining them. Don't force-push or throw away
your local work to get past the error.

Reference: [Git pull](https://git-scm.com/docs/git-pull).

### A small command reference

| Question or action | Command |
| --- | --- |
| Where am I? | `pwd` |
| What changed, and what is staged? | `git status` |
| What changed in tracked files since staging? | `git diff` |
| What will the next commit contain? | `git diff --staged` |
| Stage a particular file | `git add filename` |
| Record the staged changes | `git commit -m "Explain the change"` |
| Read the history and branch labels | `git log --oneline --graph --all --decorate` |
| Create and switch to a branch | `git switch -c branch-name` |
| Switch to an existing branch | `git switch main` |
| Merge into the current branch | `git merge branch-name` |
| Check remote URLs | `git remote -v` |
| Download history from my fork | `git fetch origin` |
| Download new course history | `git fetch upstream` |
| Merge course updates into my current branch | `git merge --no-edit upstream/main` |
| Update from my own fork when `main` can fast-forward | `git pull --ff-only origin main` while on `main` |
| Publish `main` | `git push origin main` |
| Publish a new feature branch and track it | `git push -u origin branch-name` |

Run `git status` before and after a change. Read the diff, stage the files you
want, then check the staged diff before committing. Push when you're ready to
share the commits. If Git stops with an error, read what it says before trying
another command.


### When something stops you

| What you see | What to check |
| --- | --- |
| `not a git repository` | Run `pwd`; enter the clone, not its parent directory. |
| `Author identity unknown` | Set `user.name` and `user.email` inside the repository. |
| `nothing to commit` | Save your file, then check `git status`. Did you already commit the change? |
| Authentication failed or permission denied | Check `git remote -v`, account, token expiration, selected repository, and Contents permission. |
| Push rejected as non-fast-forward | The remote has history your branch doesn't contain. Fetch, look at the graph, and show me before combining the changes. |
| Switching or merging would overwrite local changes | Look at the changes and commit the work you want to keep before switching. Don't discard it just to get past the error. |
| Conflict markers in a file | Resolve the content, remove markers, stage, inspect, and complete the merge. |

Save a notebook before staging it. Outputs and notebook metadata can add a lot
to its diff. Clear outputs you don't need to keep, and agree on who is editing
each notebook. If several people often change the same code, putting it in small
text modules can make those changes easier to compare and merge.

GitHub can display a notebook, but that preview doesn't run its Python code.
